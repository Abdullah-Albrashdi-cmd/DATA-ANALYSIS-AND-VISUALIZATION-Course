```python
#Load the dataset and the required libraries
# === Write your answer here: ===
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('student_campus_data.csv') 
```

## Task 1 — Which faculties prefer which learning modes?
The university wants to compare `Learning_Mode` preferences across the four faculties.

Create a visualization that shows the learning-mode counts **side-by-side for each faculty**. Add suitable labels, title, and legend.

```python
# === Write your answer here: ===
import numpy as np

science =  df[df["Faculty"] == "Science"]
engineering = df[df["Faculty"] == "Engineering"]
economics = df[df["Faculty"] == "Economics"]
education = df[df["Faculty"] == "Education"]

sience_count = science["Learning_Mode"].value_counts()
engineering_count = engineering["Learning_Mode"].value_counts()
economics_count = economics["Learning_Mode"].value_counts()
education_count = education["Learning_Mode"].value_counts()

categories = df["Learning_Mode"].unique()

group_A = [science_count.get(mode, 0) for mode in categories]
group_B = [engineering_count.get(mode, 0) for mode in categories]
group_C = [economics_count.get(mode, 0) for mode in categories]
group_D = [education_count.get(mode, 0) for mode in categories]

x = np.arange(len(categories))
width = 0.2

plt.bar(x - 1.5*width, group_A, width, label="Science")
plt.bar(x - 0.5*width, group_B, width, label="Engineering")
plt.bar(x + 0.5*width, group_C, width, label="Economics")
plt.bar(x + 1.5*width, group_D, width, label="Education")

plt.xticks(x, categories)
plt.legend()
plt.show()
```

## Task 2 — What share of students belongs to each year level?
Student Affairs wants one visual showing how the **entire surveyed group is divided among Year 1, Year 2, Year 3, and Year 4**.

Display the percentage for each part.

```python
# === Write your answer here: ===
year_counts = df["Year_Level"].value_counts()

plt.pie(year_counts, labels=year_counts.index, autopct='%1.1f%%')
plt.title("Distribution of Students")
plt.show()
```

## Task 3 — How many students remain above increasing quiz-score thresholds?
The course coordinator wants to see how the number of students **at or above a score threshold decreases as the threshold increases**.

Use these fixed class boundaries:

`40, 50, 60, 70, 80, 90, 100`

First calculate the required grouped frequencies, then create the appropriate cumulative-frequency graph.

```python
# === Write your answer here: ===
boundaries = [40, 50, 60, 70, 80, 90, 100]

freq = []

for boundary in boundaries:
    count = (df["Quiz_Score"] <= boundary).sum()
    freq.append(count)


plt.plot(boundaries, freq, marker='o')
plt.title("Frequency Distribution of Quiz_Score")
plt.xlabel("Quiz_Score")
plt.ylabel("Frequency")
plt.show()


```

## Task 4 — How are daily study hours distributed?
The academic support team wants to understand **how students' daily study hours are distributed and where most observations are concentrated**.

Create an appropriate visualization using **6 bins**.

```python
# === Write your answer here: ===
plt.hist(df["Study_Hours"], bins=6)

plt.xlabel("Daily Study Hours")
plt.ylabel("Frequency")
plt.title("Distribution of Daily Study Hours")
plt.show()
```

## Task 5 — Which transport mode is used most?
The university is reviewing transportation services and wants a visual that makes it easy to **compare the number of students using each transport mode and identify the largest category**.

Create the most appropriate visualization and print the most frequently used mode.

```python
# === Write your answer here: ===

transCounts = df["Transport_Mode"].value_counts()

plt.bar(transCounts.index, transCounts.values)

plt.xlabel("Transport Mode")
plt.ylabel("Number of Students")
plt.title("Students by Transport Mode")
plt.show()

most_used = transCounts.idxmax()
print("Most frequently used transport mode:", most_used)
```
