```python
random_sample2 = customers.sample(n=30, random_state=42) 
random_sample3 = customers.sample(n=100 , random_state=42) 
random_sample4 = customers.sample(n=300, random_state=42) 

print(random_sample2['Satisfaction_Score'].mean())
print(random_sample3['Satisfaction_Score'].mean())
print(random_sample4['Satisfaction_Score'].mean())

# larg sample generally give more stable estimate of the population mean
```

```python
nizwa = customers[customers["Branch"] == "Nizwa"] 
seeb = customers[customers["Branch"] == "Seeb"] 
salalah = customers[customers["Branch"] == "Salalah"] 
sohar = customers[customers["Branch"] == "Sohar"] 
muscat_city = customers[customers["Branch"] == "Muscat City"] 
sur = customers[customers["Branch"] == "Sur"]
```

```python
nizwa_sample = nizwa.sample(n=15, random_state=42)
seeb_sample = seeb.sample(n=15, random_state=42)
salalah_sample = salalah.sample(n=15, random_state=42)
sohar_sample = sohar.sample(n=15, random_state=42)
muscat_city_sample = muscat_city.sample(n=15, random_state=42)
sur_sample = sur.sample(n=15, random_state=42)
```

```python
stratified_sample = pd.concat([nizwa_sample, seeb_sample, salalah_sample, sohar_sample, muscat_city_sample, sur_sample])
```

```python
print(stratified_sample['Branch'].value_counts())
```

```python
stratified_mean = stratified_sample["Satisfaction_Score"].mean()
print("Stratified sample mean:", stratified_mean)
print("Population mean:", customers['Satisfaction_Score'].mean())
```
