```python
import matplotlib.pyplot as plt 
```

```python
plt.scatter(sales['Units'], sales['Revenue']) 
plt.xlabel('Units Sold') 
plt.ylabel('Revenue (OMR)') 
plt.title('Units Sold vs Revenue') 
plt.show() 
```

```python
print(sales.loc[sales["Revenue"].idxmax(),["Order_ID", "Region", "Product", "Units", "Revenue"]])
```

```python
print(sales.groupby('Region')['Revenue'].sum())

sales.groupby('Region')['Revenue'].sum().plot(kind='bar')

plt.xlabel('Region')
plt.ylabel('Total Revenue (OMR)')
plt.title('Total Revenue by Region')

plt.show()
```
